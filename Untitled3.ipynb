```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import  train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score

data={'House (sqrt)': [200,300,400,500,600,700,800],
      'Price (Lakhs)':[25,32,40,50,65,80,95]}
df=pd.DataFrame(data)
x = df[['House (sqrt)']]
y = df['Price (Lakhs)']

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3)

model=LinearRegression()


model.fit(x_train,y_train)
print("prediction of new value is: ",model.predict([[100]]),'Lack')

MSE = mean_squared_error(y_test,x_test)
print('MSE value is: ',mean_squared_error(y_test,x_test))
print('R2 value is: ',r2_score(y_test,x_test))

plt.figure(figsize=(10,6))
plt.scatter(x,y,color="red",marker="o",edgecolor="black",alpha=1)
plt.plot(x,model.predict(x),label = 'Regression line',color="red")
plt.xlabel('House (sqrt)')
plt.ylabel('Price (Lakhs)')
plt.grid(linestyle='--')
plt.legend()
plt.show()
```
